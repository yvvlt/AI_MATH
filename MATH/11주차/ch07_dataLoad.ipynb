```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.genfromtxt('housing.csv')
x = data[:, 5] # 6번째 RM
yt = data[:,13]     # 14번째 PRICE
print(x.shape)
print(yt.shape)
```

산점도 출력

```python
# 산점도 출력1
plt.scatter(x, yt)
plt.xlabel('RM')
plt.ylabel('PRICE')
plt.show()
```

```python
# 산점도 출력2
plt.figure(figsize=(6, 6))
plt.scatter(x, yt, s=15, c='b', marker='o')
plt.xlabel('RM', fontsize=14)
plt.ylabel('PRICE', fontsize=14)
plt.show()
```
